## Per language

```python
# Per-language recall@1 
language_ranges = {
    "de": slice(0, 1000),
    "en": slice(1000, 2000),
    "es": slice(2000, 3000),
    "fi": slice(3000, 4000),
    "ja": slice(4000, 5000),
    "ko": slice(5000, 6000),
    "ru": slice(6000, 7000),
    "th": slice(7000, 8000),
    "tr": slice(8000, 9000),
    "zh": slice(9000, 10000),
}

seed_to_data = {
    33: sapbert_data_33,
    42: sapbert_data_42,
    100: sapbert_data_100,
}

seed_to_recall1 = {}
for seed, data in seed_to_data.items():
    seed_to_recall1[seed] = {}
    for lang, lang_slice in language_ranges.items():
        _, recalls = eval_recall_sapbert(
            data[lang_slice],
            ranks=(1,),
            top_k=256,
        )
        seed_to_recall1[seed][lang] = recalls.get(1, 0.0)

print("Seed 33 recall@1:", seed_to_recall1[33])
print("Seed 42 recall@1:", seed_to_recall1[42])
print("Seed 100 recall@1:", seed_to_recall1[100])
```

```python
import pandas as pd

summary_rows = []
for lang in language_ranges:
    vals = [seed_to_recall1[seed][lang] for seed in sorted(seed_to_recall1)]
    summary_rows.append({
        "language": lang,
        "seed_33": vals[0],
        "seed_42": vals[1],
        "seed_100": vals[2],
        "mean": sum(vals) / len(vals),
        "std": pd.Series(vals).std(ddof=1),
    })

summary_df = pd.DataFrame(summary_rows)
summary_df = summary_df[["language", "seed_33", "seed_42", "seed_100", "mean", "std"]]
print(summary_df.to_string(index=False))
```
